```python
from langchain_community.document_loaders.pdf import PyPDFLoader

from pathlib import Path

from pprint import pp

import os

```

```python
print("Current folder:", os.getcwd())

file_path = Path.cwd() / "attention_is_all_you_need.pdf"
print(file_path.exists())

file_path.as_posix()
```

```python
loader = PyPDFLoader(file_path=file_path.as_posix(),mode='page')

documents = loader.load()

print(documents)
```

```python

len(documents)
pp(documents)

print(documents[0].metadata)
```

Image PDF loader

```python
from langchain_community.document_loaders.pdf import PDFMinerLoader
from langchain_community.document_loaders.parsers import RapidOCRBlobParser

image_loader = PDFMinerLoader(file_path=file_path.as_posix(),mode='page',extract_images=True,images_parser=RapidOCRBlobParser(),images_inner_format="html-img")

documents_with_images = image_loader.load()

print(documents_with_images)
```
